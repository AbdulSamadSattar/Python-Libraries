# NumPy: Quick Overview

**NumPy** (Numerical Python) is the foundational library for scientific computing in Python. It provides high-performance multidimensional array objects and tools for working with them.

---

### Key Data Structure
* **ndarray:** A fast, memory-efficient $n$-dimensional array providing vectorized operations and broadcasting capabilities.

---

### Key Features
* **Vectorized Operations:** Performs element-wise mathematical calculations without explicit Python `for` loops.
* **Broadcasting:** Allows arithmetic operations on arrays of different shapes automatically.
* **Linear Algebra & Math:** Built-in functions for matrix multiplication, Fourier transforms, random number generation, and statistics.
* **C/C++ Integration:** Serves as the high-performance underlying engine for libraries like Pandas, SciPy, and Scikit-Learn.

# Empty like
np.empty_like(arr) creates a new array with the same shape and dtype as arr, but filled with uninitialized garbage values (whatever was already in memory) — it does not set them to zero.
Why use it? It's faster than np.zeros_like() or np.ones_like() because it skips initializing the values — useful when you're about to immediately overwrite every element anyway (e.g., in a loop).

```python
a = np.array([[1, 2, 3],
              [4, 5, 6]])

b = np.empty_like(a)

print(b.shape)   # (2, 3)  -> same shape as a
print(b.dtype)   # int64   -> same dtype as a
print(b)          # random/garbage values, NOT zeros
```

```python
result = np.empty_like(a)
for i in range(a.shape[0]):
    result[i] = a[i] * 2   # filling it right after creation

print(result)
```
