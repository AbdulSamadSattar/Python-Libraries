# Matplotlib: Quick Overview

**Matplotlib** is the foundational 2D plotting library for Python. It provides a comprehensive framework for creating static, animated, and interactive visualizations.

---

### Key Features
* **Pyplot Interface:** Offers a MATLAB-like state-based interface (`matplotlib.pyplot`) for fast and easy plotting.
* **Customization:** Fine-grained control over colors, labels, axes, fonts, and layout.
* **Ecosystem Integration:** Works seamlessly with `NumPy`, `Pandas`, and `SciPy`.
* **Export Options:** Supports high-quality exports to vector and raster formats (PNG, PDF, SVG, EPS).

---

* **macOS / Linux:**
```bash

```



**3. Install Matplotlib**


````python
import matplotlib.pyplot as plt
````

````python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
````

# Bar Graph

````python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
c = ['c', 'm', 'y', 'k', 'b']
plt.bar(x, y, color=c)
plt.title("Bar Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis") 
plt.show()
````

### Hexadecimal Color Code

A hexadecimal color code uses a 6-digit representation based on three 2-digit RGB (Red, Green, Blue) values in base-16 arithmetic.

---

### Component Breakdown

| Symbol / Pair | Color Channel | Hex Value | Decimal Equivalent (0-255) | Intensity Level |
| :--- | :--- | :--- | :--- | :--- |
| **`#`** | Prefix | — | — | Indicates Hex Notation |
| **`05`** | Red | `05` | $5 / 255$ | Extremely Low ($\sim 2\%$) |
| **`04`** | Green | `04` | $4 / 255$ | Extremely Low ($\sim 1.5\%$) |
| **`aa`** | Blue | `AA` | $170 / 255$ | Moderate High ($\sim 66.7\%$) |

````python
import matplotlib.pyplot as plt

# Data prepare karna (Maan lijiye hum Titanic ke 'Sex' column ka distribution dekh
counts = [577, 314] # Male aur Female ka count
labels = ['Male', 'Female']
colors = ['skyblue', 'lightpink']
explode = (0.1, 0) # 'Male' wale hisse ko thoda bahar nikalne ke liye

# Plotting
plt.figure(figsize=(6, 6))
plt.pie(counts, labels=labels, autopct='%1.1f%%', startangle=140, colors=colors, explode=explode, shadow=True)

# Title dena
plt.title('Gender Distribution in Titanic')

# Chart dikhana
plt.show()
````

````python
fig, ax = plt.subplots()

circle = plt.Circle((5, 5), 5, fill=False, color='b', linewidth=2)
ax.add_patch(circle)

ax.axhline(0, color='k')
ax.axvline(0, color='k')

ax.set_xlim(-1, 12)
ax.set_ylim(-1, 12)
ax.set_aspect('equal', adjustable='box')
ax.set_title("Circle of radius 5 cm tangent to x and y axes")

plt.show()
````
