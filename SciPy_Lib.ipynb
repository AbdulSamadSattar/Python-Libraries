# SciPy Library

## 📌 Introduction

**SciPy (Scientific Python)** is an open-source Python library used for **scientific and technical computing**. It provides advanced mathematical and scientific functions that build on the capabilities of **NumPy**.

SciPy is commonly used for **optimization, statistics, linear algebra, integration, signal processing, image processing, and scientific calculations**.

## 🎯 Key Features

* **Optimization** – finding minimum or maximum values of functions.
* **Linear Algebra** – solving equations and performing matrix operations.
* **Integration** – calculating definite and indefinite integrals.
* **Statistics** – probability distributions and statistical tests.
* **Signal Processing** – filtering and analyzing signals.
* **Spatial Processing** – working with distances, geometry, and spatial data.
* **Interpolation** – estimating values between known data points.


> **In short:** NumPy provides the foundation for numerical computing in Python, while SciPy provides additional advanced tools for scientific and technical computing.


```python
from scipy import stats as st
```

```python
data = [85, 90, 78, 92, 88, 95, 80, 85, 89] #not for multimode

mode_result = st.mode(data)
print(f"Mode using SciPy : {mode_result.mode}")
```
