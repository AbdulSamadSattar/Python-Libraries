# Seaborn: Quick Overview

**Seaborn** is a Python data visualization library built on top of Matplotlib. It provides a high-level interface for drawing attractive, informative, and statistical graphics.

---

### Key Features
* **Built-in Themes:** Comes with polished default styles and color palettes.
* **Dataset-Oriented:** Integrates directly with `Pandas` DataFrames.
* **Statistical Plotting:** Simplifies complex visualizations like heatmaps, pair plots, and regression lines.
* **Matplotlib Backed:** Allows full customization using standard Matplotlib commands.

---

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
var1 = [1,2,3,4,5]
var2 = [5,4,3,2,1]
sns.lineplot(x=var1, y=var2)
plt.title("Line Plot using Seaborn")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
df1 = sns.load_dataset('penguins')
df1
```

```python
sns.lineplot(x='bill_length_mm', y='flipper_length_mm', data=df1)
```

```python
sns.lineplot(x='bill_length_mm', y='flipper_length_mm', data=df1, hue = 'sex')
plt.show()
```

```python
sns.lineplot(x='bill_length_mm', y='flipper_length_mm', data=df1, hue = 'sex', markers = ['o','>'])
# sns.lineplot(x='bill_length_mm', y='flipper_length_mm', data=df1, hue = 'sex', palette='rocker_r', markers = ['o','>'])
plt.show()
```
